# FloodScope 3D: map real water from Sentinel-2 with Prithvi (Sen1Floods11)

Runs IBM/NASA's **Prithvi-EO-2.0-300M-TL-Sen1Floods11** (Hugging Face, 1.3 GB checkpoint) on the least cloudy Sentinel-2 L2A scene over a point, using IBM/NASA's own `inference.py`.

The result (water mask, transparent PNG overlay, `meta.json`) goes in `satellite_results/`. Copy that folder into the repo at `ml/satellite_results/` and the 3D view can overlay it to compare with the simulated flood.

Limits: optical satellites cannot see through cloud, so floods under monsoon cloud are often invisible. The model was fine-tuned on Sentinel-2 L1C; this uses L2A (a modest domain shift).
